```python
import sqlite3
import pandas as pd

# Connect to the database
conn = sqlite3.connect(r"C:\Users\user\Downloads\chinook.db")
cursor = conn.cursor()
```

```python
### 🎯 Query 1: Top 10 Customers by Total Spending

# This query joins the `invoices` and `customers` tables to calculate how much each customer has spent in total. 
# The result is sorted in descending order to find the top 10.
```

```python
# TOP 10 Customers with highest total spending

query = '''SELECT customers.FirstName ||' '|| customers.LastName AS Customer, SUM(invoices.Total) AS TotalSpent
FROM invoices
JOIN customers ON invoices.CustomerId = customers.CustomerId
GROUP BY Customer
ORDER BY TotalSpent DESC
LIMIT 10;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 2: Top 10 artists by sales

# This query joins the `invoice_items`, `tracks`, `albums` and `artists` tables to calculate total sales of each artist's tracks. 
# The result is sorted in descending order to find the top 10.
```

```python
# Top 10 artists by sales

query = '''SELECT artists.Name AS artist,
		SUM(invoice_items.UnitPrice) AS TotalSale
FROM invoice_items
JOIN tracks ON invoice_items.TrackId = tracks.TrackId
JOIN albums ON tracks.AlbumId = albums.AlbumId
JOIN artists ON albums.ArtistId = artists.ArtistId
GROUP BY artist
ORDER BY TotalSale DESC
LIMIT 10;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 3: Most popular genres

# This query joins the `genres`, `tracks` and `invoice_items` tables to calculate sales count by genre. 
# The result is sorted in descending order to find the top 10.
```

```python
# Most popular genres

query = '''SELECT genres.Name AS genre,
		COUNT(invoice_items.TrackId) AS TotalCount
FROM genres
JOIN tracks ON genres.GenreId = tracks.GenreId
JOIN invoice_items ON tracks.TrackId = invoice_items.TrackId
GROUP BY genre
ORDER BY TotalCount DESC
LIMIT 10;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 4: Best-selling albums

# This query joins the `albums`, `tracks` and `invoice_items` tables to calculate total sales count for each album. 
# The result is sorted in descending order to find the top 5.
```

```python
# Best-selling albums

query = '''SELECT albums.Title AS album,
		COUNT(invoice_items.TrackId) AS TotalCount
FROM albums
JOIN tracks ON albums.AlbumId = tracks.AlbumId
JOIN invoice_items ON tracks.TrackId = invoice_items.TrackId
GROUP BY album
ORDER BY TotalCount DESC
LIMIT 5;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 5: Top 10 countries by sales

# This query uses the `invoices` table to calculate total sales count for each country. 
# The result is sorted in descending order to find the top 10.
```

```python
# Top 10 countries by sales

query = '''SELECT invoices.BillingCountry AS country,
		SUM(invoices.Total) AS sales
FROM invoices
GROUP BY country
ORDER BY sales DESC
LIMIT 10;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 6: Monthly revenue trend

# This query uses the `invoices` table to calculate sales per month. 
```

```python
# Monthly revenue trend

query = '''SELECT 
	strftime('%Y-%m', invoices.InvoiceDate) AS Month,
	SUM(invoices.Total) AS MonthlyRevenue
FROM invoices
GROUP BY Month
ORDER BY Month;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```

```python
### 🎯 Query 7: Top 5 employees by sales

# This query joins the `employees`, `customers` and `invoices` tables to calculate the total sales made by each employee. 
# The result is sorted in descending order to find the top 5.
```

```python
# Top 5 employees by sales

query = '''SELECT employees.FirstName ||' '|| employees.LastName AS employee,
		SUM(invoices.Total) AS TotalSales
FROM employees
LEFT JOIN customers ON employees.EmployeeId = customers.SupportRepId
LEFT JOIN invoices ON customers.CustomerId = invoices.CustomerId
GROUP BY employee
ORDER BY TotalSales DESC
LIMIT 5;'''

# Fetch results

df = pd.read_sql_query(query, conn)
df
```
